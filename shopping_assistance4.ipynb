```python
# ============================================================
# AI SHOPPING RESEARCH - MULTI AGENT SYSTEM
# ============================================================

!pip install -q gradio


import gradio as gr
import re
from datetime import datetime


# ============================================================
# PRODUCT DATABASE
# ============================================================

PRODUCTS = [

    # ---------------- LAPTOPS ----------------
    {
        "name": "HP 15 Laptop",
        "category": "laptop",
        "price": 45999,
        "brand": "HP",
        "specifications": "Intel Core i5, 8GB RAM, 512GB SSD, 15.6 inch",
        "rating": 4.3
    },

    {
        "name": "Lenovo IdeaPad Slim 3",
        "category": "laptop",
        "price": 42999,
        "brand": "Lenovo",
        "specifications": "AMD Ryzen 5, 8GB RAM, 512GB SSD, 15.6 inch",
        "rating": 4.2
    },

    {
        "name": "ASUS Vivobook 15",
        "category": "laptop",
        "price": 48999,
        "brand": "ASUS",
        "specifications": "Intel Core i5, 16GB RAM, 512GB SSD, 15.6 inch",
        "rating": 4.4
    },


    # ---------------- PHONES ----------------
    {
        "name": "Samsung Galaxy M35",
        "category": "phone",
        "price": 16999,
        "brand": "Samsung",
        "specifications": "AMOLED, 6000mAh Battery, 50MP Camera, 5G",
        "rating": 4.4
    },

    {
        "name": "OnePlus Nord CE",
        "category": "phone",
        "price": 19999,
        "brand": "OnePlus",
        "specifications": "AMOLED, 5500mAh Battery, 50MP Camera, 5G",
        "rating": 4.3
    },

    {
        "name": "Redmi Note Series",
        "category": "phone",
        "price": 14999,
        "brand": "Redmi",
        "specifications": "AMOLED, 5000mAh Battery, 108MP Camera, 5G",
        "rating": 4.2
    },


    # ---------------- HEADPHONES ----------------
    {
        "name": "Sony WH Series",
        "category": "headphones",
        "price": 4999,
        "brand": "Sony",
        "specifications": "Wireless, ANC, Bluetooth, 30-hour Battery",
        "rating": 4.5
    },

    {
        "name": "JBL Tune Series",
        "category": "headphones",
        "price": 2999,
        "brand": "JBL",
        "specifications": "Wireless, Bluetooth, 40-hour Battery",
        "rating": 4.4
    },


    # ---------------- SMARTWATCHES ----------------
    {
        "name": "Noise ColorFit",
        "category": "smartwatch",
        "price": 2499,
        "brand": "Noise",
        "specifications": "AMOLED Display, Heart Rate, SpO2, Bluetooth",
        "rating": 4.2
    },

    {
        "name": "boAt Wave",
        "category": "smartwatch",
        "price": 1999,
        "brand": "boAt",
        "specifications": "1.69 inch Display, Fitness Tracking, Bluetooth",
        "rating": 4.1
    },


    # ---------------- SHOES ----------------
    {
        "name": "Nike Running Shoes",
        "category": "shoes",
        "price": 2999,
        "brand": "Nike",
        "specifications": "Running, Lightweight, Breathable",
        "rating": 4.5
    },

    {
        "name": "Adidas Sports Shoes",
        "category": "shoes",
        "price": 3499,
        "brand": "Adidas",
        "specifications": "Sports, Lightweight, Comfortable",
        "rating": 4.4
    },


    # ---------------- DRESSES ----------------
    {
        "name": "Women Floral Dress",
        "category": "dress",
        "price": 1299,
        "brand": "Fashion Brand",
        "specifications": "Floral Print, Casual, Cotton",
        "rating": 4.2
    },

    {
        "name": "Women Party Dress",
        "category": "dress",
        "price": 2499,
        "brand": "Style Brand",
        "specifications": "Party Wear, Polyester, Modern Design",
        "rating": 4.3
    }
]


# ============================================================
# AGENT 1 - RESEARCH AGENT
# ============================================================

def research_agent(query):

    query = query.lower().strip()

    # Find numbers from query
    numbers = re.findall(r"\d+", query)

    budget = None

    if numbers:
        budget = int(numbers[-1])


    # Product category detection
    categories = {
        "laptop": ["laptop", "notebook", "computer"],
        "phone": ["phone", "mobile", "smartphone"],
        "headphones": ["headphone", "headphones", "earphone", "earbuds"],
        "smartwatch": ["smartwatch", "smart watch"],
        "shoes": ["shoe", "shoes", "sneakers"],
        "dress": ["dress", "gown"]
    }


    category = None


    for category_name, keywords in categories.items():

        for keyword in keywords:

            if keyword in query:

                category = category_name
                break

        if category is not None:
            break


    # Search products
    results = []


    for product in PRODUCTS:

        # Check category
        if category is not None:

            if product["category"] != category:
                continue


        # Check budget
        if budget is not None:

            if product["price"] > budget:
                continue


        results.append(product)


    return results, category, budget


# ============================================================
# AGENT 2 - COMPARISON AGENT
# ============================================================

def comparison_agent(products):

    if len(products) == 0:

        return "❌ No products found."


    # Sort by rating
    sorted_products = sorted(
        products,
        key=lambda x: x["rating"],
        reverse=True
    )


    result = ""


    for index, product in enumerate(sorted_products, 1):

        result += f"""
### {index}. {product["name"]}

💰 **Price:** ₹{product["price"]:,}

🏷️ **Brand:** {product["brand"]}

⭐ **Rating:** {product["rating"]}/5

📋 **Specifications:**
{product["specifications"]}

---
"""


    return result


# ============================================================
# AGENT 3 - ANALYSIS AGENT
# ============================================================

def analysis_agent(products):

    if len(products) == 0:

        return "No products available for analysis."


    highest_rated = max(
        products,
        key=lambda x: x["rating"]
    )


    lowest_price = min(
        products,
        key=lambda x: x["price"]
    )


    analysis = f"""
### 📊 Product Analysis

**Highest Rated:**

{highest_rated["name"]}

⭐ Rating: {highest_rated["rating"]}/5


**Lowest Price:**

{lowest_price["name"]}

💰 Price: ₹{lowest_price["price"]:,}


**Products Found:** {len(products)}


The AI analysis considers:

• Product price
• Customer rating
• Brand
• Specifications
• Overall value
"""


    return analysis


# ============================================================
# AGENT 4 - RECOMMENDATION AGENT
# ============================================================

def recommendation_agent(products):

    if len(products) == 0:

        return "No recommendation available."


    max_price = max(
        product["price"]
        for product in products
    )


    scored_products = []


    for product in products:

        # Rating score
        rating_score = product["rating"] / 5


        # Price score
        price_score = 1 - (
            product["price"] / max_price
        )


        # Overall score
        score = (
            rating_score * 0.6
            +
            price_score * 0.4
        )


        scored_products.append(
            (product, score)
        )


    # Sort products
    scored_products.sort(
        key=lambda x: x[1],
        reverse=True
    )


    recommended_product = scored_products[0][0]


    result = f"""
### 🏆 AI Recommendation

## {recommended_product["name"]}

💰 **Price:** ₹{recommended_product["price"]:,}

⭐ **Rating:** {recommended_product["rating"]}/5

🏷️ **Brand:** {recommended_product["brand"]}

📋 **Specifications:**

{recommended_product["specifications"]}


### Why this product?

The recommendation is based on:

✅ Price
✅ Rating
✅ Specifications
✅ Brand
✅ Overall value
"""


    return result


# ============================================================
# MAIN MULTI-AGENT SYSTEM
# ============================================================

def shopping_research(query):

    if not query or not query.strip():

        return "⚠️ Please enter a product."


    # Agent 1
    products, category, budget = research_agent(query)


    # Agent 2
    comparison = comparison_agent(products)


    # Agent 3
    analysis = analysis_agent(products)


    # Agent 4
    recommendation = recommendation_agent(products)


    # Category display
    if category is None:
        category_text = "All Products"
    else:
        category_text = category.title()


    # Budget display
    if budget is None:
        budget_text = "Not specified"
    else:
        budget_text = f"₹{budget:,}"


    # Final report
    report = f"""
# 🔎 AI Research Report

### Search Query

**{query}**


### Product Category

**{category_text}**


### Budget

**{budget_text}**

---

# 🔍 Research Agent

The Research Agent searched the product database
according to the requested category and budget.

---

# 📋 Product Comparison

{comparison}

---

# 📊 Analysis Agent

{analysis}

---

# 🤖 Recommendation Agent

{recommendation}

---

# 🔄 Multi-Agent Workflow

**User Query**

↓

**Research Agent**

↓

**Comparison Agent**

↓

**Analysis Agent**

↓

**Recommendation Agent**

↓

**Final Shopping Report**

---

_Report generated: {datetime.now().strftime("%d-%m-%Y %H:%M:%S")}_
"""


    return report


# ============================================================
# GRADIO INTERFACE
# ============================================================

with gr.Blocks(
    title="AI Shopping Research"
) as demo:


    gr.Markdown(
        """
# 🛍️ AI Shopping Research

### Research • Compare • Analyze • Recommend
"""
    )


    product_input = gr.Textbox(
        label="Enter Product",
        placeholder="Example: laptop under 50000",
        value="laptop under 50000",
        lines=1
    )


    research_button = gr.Button(
        "🔎 Start AI Research",
        variant="primary"
    )


    output = gr.Markdown()


    research_button.click(
        fn=shopping_research,
        inputs=product_input,
        outputs=output
    )


# ============================================================
# RUN APPLICATION
# ============================================================

demo.launch(
    share=True,
    debug=True
)
```
